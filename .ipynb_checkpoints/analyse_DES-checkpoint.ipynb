## Étape 1 : Fonctions de conversion bits ↔ entiers

```python
def int_vers_bits(n, longueur):
    
    bits = []
    for i in range(longueur-1, -1, -1):
        bits.append((n >> i) & 1)
    return bits

def bits_vers_int(bits):
    
    n = 0
    for bit in bits:
        n = (n << 1) | bit
    return n
```

## Étape 2 : Fonction de permutation

```python
def permuter(bits, table):
    
    return [bits[i-1] for i in table]
```

## Étape 3 : Génération des 16 sous-clés

```python
def rotation_gauche(bits, n):
    """
    Rotation vers la gauche de n positions
    Exemple : rotation_gauche([1,0,0,1,1], 1) → [0,0,1,1,1]
    """
    return bits[n:] + bits[:n]


def generer_sous_cles(cle_64bits):
    
    # Étape 1 : PC-1 (64 → 56 bits)
    cle_56bits = permuter(cle_64bits, PC1)
    
    # Étape 2 : Diviser en C et D de 28 bits
    C = cle_56bits[:28]
    D = cle_56bits[28:]
    
    # Étape 3 : Générer les 16 sous-clés
    sous_cles = []
    for i in range(16):
        # Rotation de C et D
        C = rotation_gauche(C, DECALAGES[i])
        D = rotation_gauche(D, DECALAGES[i])
        
        # PC-2 (56 → 48 bits)
        CD = C + D
        Ki = permuter(CD, PC2)
        sous_cles.append(Ki)
    
    return sous_cles
```

## Étape 4 : La fonction F (fonction de tour)

```python
def expansion_E(R):
    """
    Expansion : 32 bits → 48 bits
    Certains bits sont dupliqués selon la table E
    """
    return permuter(R, E)


def boites_S(bits_48):
    """
    8 boîtes S : 48 bits → 32 bits
    
    Divise les 48 bits en 8 groupes de 6 bits
    Chaque groupe entre dans une boîte S → 4 bits en sortie
    """
    resultat = []
    
    for i in range(8):
        # Extraire les 6 bits du groupe i
        groupe = bits_48[i*6 : i*6+6]
        
        # Calculer la ligne (bit 1 et bit 6)
        ligne = (groupe[0] << 1) | groupe[5]
        
        # Calculer la colonne (bits 2 à 5)
        colonne = (groupe[1] << 3) | (groupe[2] << 2) | (groupe[3] << 1) | groupe[4]
        
        # Lire la valeur dans la boîte S
        valeur = S_BOXES[i][ligne][colonne]
        
        # Convertir en 4 bits
        resultat += int_vers_bits(valeur, 4)
    
    return resultat


def fonction_F(R, K):
    
    # Étape 1 : Expansion E (32 → 48 bits)
    R_etendu = expansion_E(R)
    
    # Étape 2 : XOR avec la sous-clé K
    xor_result = [R_etendu[i] ^ K[i] for i in range(48)]
    
    # Étape 3 : Boîtes S (48 → 32 bits)
    apres_S = boites_S(xor_result)
    
    # Étape 4 : Permutation P (32 → 32 bits)
    resultat = permuter(apres_S, P)
    
    return resultat
```

```python
def DES_chiffrer(message_hex, cle_hex):
    """
    Chiffrement DES complet
    
    message_hex : liste de 8 octets en hexa [0x86, 0x67, ...]
    cle_hex     : liste de 8 octets en hexa
    retourne    : liste de 8 octets chiffrés en hexa
    """
    # Étape 1 : Convertir message et clé en bits
    message_bits = []
    for octet in message_hex:
        message_bits += int_vers_bits(octet, 8)
    
    cle_bits = []
    for octet in cle_hex:
        cle_bits += int_vers_bits(octet, 8)
    
    # Étape 2 : Permutation initiale IP
    message_ip = permuter(message_bits, IP)
    
    # Étape 3 : Diviser en L0 et R0
    L = message_ip[:32]
    R = message_ip[32:]
    
    # Étape 4 : Générer les 16 sous-clés
    sous_cles = generer_sous_cles(cle_bits)
    
    # Étape 5 : 16 tours Feistel
    for i in range(16):
        L_nouveau = R
        R_nouveau = [L[j] ^ fonction_F(R, sous_cles[i])[j] for j in range(32)]
        L = L_nouveau
        R = R_nouveau
    
    # Étape 6 : Permutation finale IP_INV
    # Attention : on met R avant L !
    chiffre_bits = permuter(R + L, IP_INV)
    
    # Étape 7 : Convertir en octets hexa
    chiffre_hex = []
    for i in range(8):
        octet_bits = chiffre_bits[i*8 : i*8+8]
        chiffre_hex.append(bits_vers_int(octet_bits))
    
    return chiffre_hex
```

```python
# Notre message clair
message = [0x86, 0x67, 0xA7, 0x0B, 0x08, 0xE6, 0x5B, 0x61]

# Chiffré juste attendu
chiffre_attendu = [0x61, 0xD5, 0x83, 0x36, 0xBF, 0xD4, 0x83, 0xB0]

# Test avec une clé nulle pour vérifier que le code tourne
cle_test = [0x00]*8
chiffre = DES_chiffrer(message, cle_test)
print("Message clair  :", [hex(x) for x in message])
print("Chiffré obtenu :", [hex(x) for x in chiffre])
print("Chiffré attendu:", [hex(x) for x in chiffre_attendu])
```
