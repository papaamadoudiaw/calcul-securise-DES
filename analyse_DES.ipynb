## Étape 1 : Fonctions de conversion bits ↔ entiers

```python
def int_vers_bits(n, longueur):
    
    bits = []
    for i in range(longueur-1, -1, -1):
        bits.append((n >> i) & 1)
    return bits

def bits_vers_int(bits):
    
    n = 0
    for bit in bits:
        n = (n << 1) | bit
    return n
```

## Étape 2 : Fonction de permutation

```python
def permuter(bits, table):
    
    return [bits[i-1] for i in table]
```

## Étape 3 : Génération des 16 sous-clés

```python
def rotation_gauche(bits, n):
    """
    Rotation vers la gauche de n positions
    Exemple : rotation_gauche([1,0,0,1,1], 1) → [0,0,1,1,1]
    """
    return bits[n:] + bits[:n]


def generer_sous_cles(cle_64bits):
    
    # Étape 1 : PC-1 (64 → 56 bits)
    cle_56bits = permuter(cle_64bits, PC1)
    
    # Étape 2 : Diviser en C et D de 28 bits
    C = cle_56bits[:28]
    D = cle_56bits[28:]
    
    # Étape 3 : Générer les 16 sous-clés
    sous_cles = []
    for i in range(16):
        # Rotation de C et D
        C = rotation_gauche(C, DECALAGES[i])
        D = rotation_gauche(D, DECALAGES[i])
        
        # PC-2 (56 → 48 bits)
        CD = C + D
        Ki = permuter(CD, PC2)
        sous_cles.append(Ki)
    
    return sous_cles
```

## Étape 4 : La fonction F (fonction de tour)

```python
def expansion_E(R):
    """
    Expansion : 32 bits → 48 bits
    Certains bits sont dupliqués selon la table E
    """
    return permuter(R, E)


def boites_S(bits_48):
    """
    8 boîtes S : 48 bits → 32 bits
    
    Divise les 48 bits en 8 groupes de 6 bits
    Chaque groupe entre dans une boîte S → 4 bits en sortie
    """
    resultat = []
    
    for i in range(8):
        # Extraire les 6 bits du groupe i
        groupe = bits_48[i*6 : i*6+6]
        
        # Calculer la ligne (bit 1 et bit 6)
        ligne = (groupe[0] << 1) | groupe[5]
        
        # Calculer la colonne (bits 2 à 5)
        colonne = (groupe[1] << 3) | (groupe[2] << 2) | (groupe[3] << 1) | groupe[4]
        
        # Lire la valeur dans la boîte S
        valeur = S_BOXES[i][ligne][colonne]
        
        # Convertir en 4 bits
        resultat += int_vers_bits(valeur, 4)
    
    return resultat


def fonction_F(R, K):
    
    # Étape 1 : Expansion E (32 → 48 bits)
    R_etendu = expansion_E(R)
    
    # Étape 2 : XOR avec la sous-clé K
    xor_result = [R_etendu[i] ^ K[i] for i in range(48)]
    
    # Étape 3 : Boîtes S (48 → 32 bits)
    apres_S = boites_S(xor_result)
    
    # Étape 4 : Permutation P (32 → 32 bits)
    resultat = permuter(apres_S, P)
    
    return resultat
```

```python
def DES_chiffrer(message_hex, cle_hex):
    """
    Chiffrement DES complet
    
    message_hex : liste de 8 octets en hexa [0x86, 0x67, ...]
    cle_hex     : liste de 8 octets en hexa
    retourne    : liste de 8 octets chiffrés en hexa
    """
    # Étape 1 : Convertir message et clé en bits
    message_bits = []
    for octet in message_hex:
        message_bits += int_vers_bits(octet, 8)
    
    cle_bits = []
    for octet in cle_hex:
        cle_bits += int_vers_bits(octet, 8)
    
    # Étape 2 : Permutation initiale IP
    message_ip = permuter(message_bits, IP)
    
    # Étape 3 : Diviser en L0 et R0
    L = message_ip[:32]
    R = message_ip[32:]
    
    # Étape 4 : Générer les 16 sous-clés
    sous_cles = generer_sous_cles(cle_bits)
    
    # Étape 5 : 16 tours Feistel
    for i in range(16):
        L_nouveau = R
        R_nouveau = [L[j] ^ fonction_F(R, sous_cles[i])[j] for j in range(32)]
        L = L_nouveau
        R = R_nouveau
    
    # Étape 6 : Permutation finale IP_INV
    # Attention : on met R avant L !
    chiffre_bits = permuter(R + L, IP_INV)
    
    # Étape 7 : Convertir en octets hexa
    chiffre_hex = []
    for i in range(8):
        octet_bits = chiffre_bits[i*8 : i*8+8]
        chiffre_hex.append(bits_vers_int(octet_bits))
    
    return chiffre_hex
```

## Étape 7 : Attaque par fautes sur le dernier tour
### Principe
- La faute est injectée sur R15
- On exploite la différence entre chiffré juste et chiffrés faux
- Pour chaque boîte S, on teste 64 valeurs possibles de sous-clé
- On élimine les valeurs incompatibles
- → On retrouve K16 (48 bits)

```python
def hex_vers_bits(liste_hex):
    """
    Convertit une liste d'octets hexa en liste de 64 bits
    Exemple : [0x61, 0xD5, ...] → [0,1,1,0,0,0,0,1, ...]
    """
    bits = []
    for octet in liste_hex:
        bits += int_vers_bits(octet, 8)
    return bits


def extraire_R15_R16(chiffre_hex):
    """
    Depuis un chiffré, extrait R15 et R16
    
    Structure du dernier tour DES :
        L16 = R15
        R16 = L15 ⊕ F(R15, K16)
    
    Après IP_INV : Chiffré = IP_INV(R16 + L16)
    Donc : IP(Chiffré) = R16 + L16
    
    → R16 = IP(Chiffré)[:32]
    → R15 = L16 = IP(Chiffré)[32:]
    """
    # Convertir en bits
    bits = hex_vers_bits(chiffre_hex)
    
    # Appliquer IP
    apres_IP = permuter(bits, IP)
    
    # Extraire R16 et L16
    R16 = apres_IP[:32]
    L16 = apres_IP[32:]
    
    # R15 = L16
    R15 = L16
    
    return R15, R16
```

## Étape 7b : Attaque sur les boîtes S

```python
def attaque_une_boite_S(numero_S, R15, R15_etoile, R16, R16_etoile):
    """
    Attaque sur la boîte S numéro numero_S (0 à 7)
    
    Principe :
    R16 ⊕ R16* = F(R15, K16) ⊕ F(R15*, K16)
    
    Pour chaque candidat k (0 à 63) :
        - Calculer sortie_S(R15,  k) 
        - Calculer sortie_S(R15*, k)
        - Si différence ≠ différence observée → éliminer k
    
    Retourne : ensemble des candidats qui survivent
    """
    # Étape 1 : Expansion de R15 et R15*
    R15_exp  = expansion_E(R15)
    R15e_exp = expansion_E(R15_etoile)
    
    # Étape 2 : Extraire les 6 bits pour la boîte S numero_S
    debut = numero_S * 6
    R15_6bits  = R15_exp [debut : debut+6]
    R15e_6bits = R15e_exp[debut : debut+6]
    
    # Étape 3 : Calculer la différence observée dans F
    # F(R15,K16) ⊕ F(R15*,K16) = R16 ⊕ R16* (car L15 s'annule)
    # Mais F contient la permutation P à la fin
    # On doit inverser P pour isoler la contribution de la boîte S i
    
    # Inverser P : trouver les 4 bits de sortie de Si dans F
    P_INV = [0] * 32
    for j in range(32):
        P_INV[P[j]-1] = j
    
    # Les 4 bits de sortie de Si correspondent aux positions
    positions_Si = [P_INV[numero_S*4 + j] for j in range(4)]
    
    # Différence observée dans R16 (après permutation P)
    diff_R16 = [R16[j] ^ R16_etoile[j] for j in range(32)]
    
    # Extraire les 4 bits correspondant à Si
    diff_Si = [diff_R16[positions_Si[j]] for j in range(4)]
    
    # Étape 4 : Tester les 64 candidats
    candidats = set(range(64))
    
    for k in range(64):
        k_bits = int_vers_bits(k, 6)
        
        # XOR avec les 6 bits de R15 et R15*
        entree_1 = [R15_6bits[j]  ^ k_bits[j] for j in range(6)]
        entree_2 = [R15e_6bits[j] ^ k_bits[j] for j in range(6)]
        
        # Calculer ligne et colonne pour S_i
        def calculer_sortie_S(entree, i):
            ligne   = (entree[0] << 1) | entree[5]
            colonne = (entree[1] << 3) | (entree[2] << 2) | (entree[3] << 1) | entree[4]
            valeur  = S_BOXES[i][ligne][colonne]
            return int_vers_bits(valeur, 4)
        
        sortie_1 = calculer_sortie_S(entree_1, numero_S)
        sortie_2 = calculer_sortie_S(entree_2, numero_S)
        
        # Différence calculée
        diff_calc = [sortie_1[j] ^ sortie_2[j] for j in range(4)]
        
        # Si incompatible → éliminer
        if diff_calc != diff_Si:
            candidats.discard(k)
    
    return candidats
```

```python
def attaque_complete(chiffre_juste, chiffres_faux):
    """
    Attaque complète sur les 8 boîtes S
    avec tous les chiffrés faux
    """
    print("=== Attaque par fautes sur le DES ===\n")
    
    # Extraire R15 et R16 depuis le chiffré juste
    R15, R16 = extraire_R15_R16(chiffre_juste)
    print(f"R15 extrait : {R15}")
    print(f"R16 extrait : {R16}\n")
    
    # Initialiser candidats : toutes les valeurs 0-63 possibles
    candidats = [set(range(64)) for _ in range(8)]
    
    # Pour chaque chiffré faux
    for idx, chiffre_faux in enumerate(chiffres_faux):
        
        # Extraire R15* et R16*
        R15_f, R16_f = extraire_R15_R16(chiffre_faux)
        
        # Si faute nulle → ignorer
        if R15_f == R15:
            print(f"Chiffré {idx+1:2d} : faute nulle → ignoré")
            continue
        
        # Attaquer chaque boîte S
        for i in range(8):
            nouveaux = attaque_une_boite_S(i, R15, R15_f, R16, R16_f)
            candidats[i] = candidats[i] & nouveaux
        
        # Afficher progression
        nb = [len(candidats[i]) for i in range(8)]
        print(f"Chiffré {idx+1:2d} : candidats = {nb}")
    
    # Résultats finaux
    print("\n=== Résultats par boîte S ===")
    K16_bits = []
    valeurs_K16 = []
    
    for i in range(8):
        nb = len(candidats[i])
        valeur = list(candidats[i])
        print(f"S{i+1} : {nb} candidat(s) → {valeur}")
        
        if nb == 1:
            v = valeur[0]
            valeurs_K16.append(v)
            K16_bits += int_vers_bits(v, 6)
        else:
            print(f"  ⚠️ Ambiguïté sur S{i+1} !")
            v = valeur[0]
            valeurs_K16.append(v)
            K16_bits += int_vers_bits(v, 6)
    
    print(f"\nK16 (48 bits) : {K16_bits}")
    print(f"Valeurs K16   : {valeurs_K16}")
    
    return K16_bits, valeurs_K16
```

```python
def inverser_PC2(K16_bits):
    """
    Retrouve les 56 bits partiels depuis K16 (48 bits)
    en inversant PC-2
    
    PC-2 sélectionne 48 bits parmi 56
    → 8 bits restent inconnus
    """
    # Initialiser 56 positions à None
    cle_56 = [None] * 56
    
    # Placer chaque bit de K16 à sa position d'origine
    for i, pos in enumerate(PC2):
        cle_56[pos-1] = K16_bits[i]
    
    # Trouver les positions manquantes
    positions_manquantes = []
    for i in range(56):
        if cle_56[i] is None:
            positions_manquantes.append(i+1)
    
    print(f"Bits connus   : {56 - len(positions_manquantes)}")
    print(f"Bits inconnus : {len(positions_manquantes)}")
    print(f"Positions manquantes : {positions_manquantes}")
    
    return cle_56, positions_manquantes
```

```python
def inverser_PC1(cle_56):
    """
    Reconstruit la clé 64 bits depuis 56 bits
    en inversant PC-1
    
    PC-1 prend 56 bits depuis la clé 64 bits
    On fait l'inverse : on replace chaque bit à sa position d'origine
    """
    cle_64_bits = [0] * 64
    
    # Replacer chaque bit à sa position d'origine dans la clé 64 bits
    for i, pos in enumerate(PC1):
        cle_64_bits[pos-1] = cle_56[i]
    
    # Convertir en liste d'octets
    cle_64_octets = []
    for i in range(8):
        octet_bits = cle_64_bits[i*8 : i*8+8]
        cle_64_octets.append(bits_vers_int(octet_bits))
    
    return cle_64_octets


def force_brute(cle_56_partielle, positions_manquantes, message, chiffre_juste):
    """
    Force brute sur les 8 bits manquants
    Teste 2^8 = 256 combinaisons
    """
    print("=== Force brute sur 8 bits manquants ===")
    print(f"Nombre de combinaisons : 2^8 = 256\n")
    
    for combinaison in range(256):
        # Convertir en 8 bits
        bits_manquants = int_vers_bits(combinaison, 8)
        
        # Remplir les positions manquantes
        cle_56 = cle_56_partielle.copy()
        for i, pos in enumerate(positions_manquantes):
            cle_56[pos-1] = bits_manquants[i]
        
        # Inverser PC-1 → clé 64 bits
        cle_64 = inverser_PC1(cle_56)
        
        # Tester le DES
        chiffre_test = DES_chiffrer(message, cle_64)
        
        # Vérifier
        if chiffre_test == chiffre_juste:
            print(f" Clé trouvée ! (combinaison {combinaison})")
            print(f"Clé 64 bits (hex) : {[hex(x) for x in cle_64]}")
            return cle_56, cle_64
    
    print("❌ Aucune clé trouvée !")
    return None, None
```

```python
def ajouter_parite(cle_64_octets):
    """
    Ajoute les bits de parité à la clé 64 bits
    Chaque octet doit avoir un nombre impair de 1
    Le bit de poids faible (bit 8) est le bit de parité
    """
    print("=== Ajout des bits de parité ===\n")
    
    cle_finale = []
    
    for i, octet in enumerate(cle_64_octets):
        # Extraire les 7 bits de données (bits 1 à 7)
        bits_7 = int_vers_bits(octet, 8)[:7]
        
        # Compter les 1
        nb_uns = sum(bits_7)
        
        # Bit de parité : rendre le nombre de 1 impair
        if nb_uns % 2 == 0:
            bit_parite = 1  # ajouter un 1 pour rendre impair
        else:
            bit_parite = 0  # garder pair → impair
        
        # Construire l'octet final
        octet_final = bits_vers_int(bits_7 + [bit_parite])
        cle_finale.append(octet_final)
        
        print(f"Octet {i+1} : {bits_7} + [{bit_parite}] "
              f"= {bin(octet_final)[2:].zfill(8)} "
              f"= {hex(octet_final)} "
              f"({nb_uns + bit_parite} uns → impair )")
    
    return cle_finale
```
